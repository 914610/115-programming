# Range

Python:
- 懶惰是一種美德
- 由0開頭
- 包含頭，不包含尾

```python
print(list(range(5)))
#區間由0開始，到5-1結束
print(list(range(0,5)))
#區間由1開始，到5-1結束
print(list(range(1,5)))
```

# Loop

Loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
list(range(10))
```

```python
#底下的程式碼會印出0-9
for i in range(10):
    print(i)
```
